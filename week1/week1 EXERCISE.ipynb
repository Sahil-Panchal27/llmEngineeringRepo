```python
# imports
import os
import json
from dotenv import load_dotenv
from openai import OpenAI 
from IPython.display import Markdown, display, update_display 
```

```python
# constants
load_dotenv(override=True)
apiKey = os.getenv('GOOGLE_API_KEY')
if apiKey:
    print(f"got api key")
else:
    print("didn't get api key")

MODEL = 'gemini-2.5-pro'
#MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(api_key=apiKey, base_url=GEMINI_BASE_URL)
```

```python
# here is the question; type over this to ask something new

question = """Explain how to get the load on a cantilever beam with uniform force distribution and a moment at the free end of the beam. Explain in formulas. The output should be in markdown.
"""
```

```python
systemPrompt = "You are a expert in a technical field. Provided a technical question, you will reply with the answer as if you are teaching a complete novice to the field." \
"your answe should contain relevant formulas. Provide your answer in markdown."
```

```python
# Get gemini to answer, with streaming
def getAnswer(question:str):
    stream = gemini.chat.completions.create(model=MODEL, messages=[
        {"role":"system","content":systemPrompt},
        {"role":"user","content":question}],
        stream=True)

    response:str = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response),display_id=display_handle.display_id)
```

```python
getAnswer(question)
```
